```python
import numpy as np
from qiskit import *
from qiskit import Aer
import matplotlib.pyplot as plt
from qiskit.visualization import plot_histogram
```

```python
# using qiskit to make bell state
circ = QuantumCircuit(2,2)
circ.initialize('00')
circ.h(0)
circ.cx(0,1)
circ.measure([0,1],[0,1])
# this creates a bell state
circ.draw('mpl')

# Use Aer's qasm_simulator
backend_sim = Aer.get_backend('qasm_simulator')

# Execute the circuit on the qasm simulator.
# We've set the number of repeats of the circuit
# to be 1024, which is the default.

num_shots = 1024
job_sim = backend_sim.run(transpile(circ, backend_sim), shots=num_shots)

# Grab the results from the job.
result_sim = job_sim.result()

counts = result_sim.get_counts(circ)
print(counts)
plot_histogram(counts)
```

```python
# drawing the circuit diagram
circ.draw('mpl')
```

```python
# try making your own 2 qubit quantum circuit in this cell
# we have the x gate, h gate , ry gate, and CNOT (cx) gates to work with
# here is an example replacing the h gate above with a x gate

# circ.x(0) applies x gate to first qubit
# circ.x(1) applies x gate to second qubit
circ = QuantumCircuit(2,2)
circ.initialize('00')
#circ.ry(np.pi/3,0)
circ.h(0)
circ.x(1)
circ.cx(0,1)
circ.x(0)
circ.h(1)
#circ.ry(np.pi/3,1)
#circ.x(0)
#circ.x(1)
circ.measure([0,1],[0,1])
# this creates a bell state
circ.draw('mpl')
num_shots = 8000
# Use Aer's qasm_simulator
backend_sim = Aer.get_backend('qasm_simulator')

# Execute the circuit on the qasm simulator.
# We've set the number of repeats of the circuit
# to be 1024, which is the default. now set to num_shots which is 2000
job_sim = backend_sim.run(transpile(circ, backend_sim), shots=num_shots)

# Grab the results from the job.
result_sim = job_sim.result()

counts = result_sim.get_counts(circ)
print(counts)
plot_histogram(counts)
```
